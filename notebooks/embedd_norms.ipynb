```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load('../Kaggle_dataset/data/token_embedding_dict.npy', allow_pickle=True).item()
norms = np.array([np.linalg.norm(point) for point in data.values()])

plt.hist(norms, bins=100)
plt.title("Histogram of norm of embedding vectors")
plt.xlabel("Norm")
plt.ylabel("Frequency")
plt.show()
```
