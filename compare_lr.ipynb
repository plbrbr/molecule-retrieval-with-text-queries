```python
import numpy as np
import plotly.graph_objects as go
```

```python
def get_losses(lines):
    losses = []
    for line in lines:
        loss = float(line.split('training loss: ')[1].split('\n')[0])
        losses.append(loss)
    return np.array(losses)

def get_scores(lines):
    scores = [0]
    for line in lines:
        score = float(line.split('score: ')[1].split('\n')[0])
        scores.append(score)
    return np.array(scores)

def plot_log_losses(all_losses, learning_rates):
    fig = go.Figure()
    for (losses, learning_rate) in zip(all_losses, learning_rates):
        fig.add_trace(go.Scatter(x=np.arange(1, 4101, 50), y=np.log(losses), mode='lines', name=f"{learning_rate:.1e}"))

    fig.update_layout(
        title='Training loss - log scale',
        xaxis_title='Iteration',
        yaxis_title='Log(loss)',
        legend_title='lr',
    )

    fig.show()

def plot_losses(all_losses, learning_rates):
    fig = go.Figure()
    for (losses, learning_rate) in zip(all_losses, learning_rates):
        fig.add_trace(go.Scatter(x=np.arange(1, 4101, 50), y=losses, mode='lines', name=f"{learning_rate:.1e}"))

    fig.update_layout(
        title='Training loss',
        xaxis_title='Iteration',
        yaxis_title='Loss',
        legend_title='lr',
    )

    fig.show()

def plot_scores(all_scores, learning_rates):
    fig = go.Figure()

    for (scores, learning_rate) in zip(all_scores, learning_rates):
        fig.add_trace(go.Scatter(x=np.arange(0, 6), y=scores, mode='lines', name=f"{learning_rate:.1e}"))
    
    fig.update_layout(
        title='Validation score',
        xaxis_title='Epoch',
        yaxis_title='Score',
        legend_title='lr',
    )

    fig.show()
```

```python
learning_rates = [5e-5, 7.5e-5, 1e-4, 5e-4, 5e-3]
filenames = [f'logs/scibert_scivocab_cased_bs_32_epochs_5_lr_{lr}.log' for lr in learning_rates]

all_losses, all_scores = [], []

fig = go.Figure()

for filename in filenames:
    with open(filename, 'r') as f:
        lines = f.readlines()
        loss_lines = [line for line in lines if 'Iteration' in line]
        score_lines = [line for line in lines if 'score' in line]
    losses = get_losses(loss_lines) 
    scores = get_scores(score_lines)
    all_losses.append(losses)
    all_scores.append(scores)

plot_losses(all_losses, learning_rates)
plot_log_losses(all_losses, learning_rates)
plot_scores(all_scores, learning_rates)
```
